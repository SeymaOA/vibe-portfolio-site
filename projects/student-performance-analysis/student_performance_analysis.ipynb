# Student Performance Analysis

This notebook explores a small student performance dataset. The goal is to practice a real data workflow: load data, inspect it, visualize patterns, and train a simple pass/fail prediction model.

## 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

sns.set_theme(style='whitegrid')
```

## 2. Load the Dataset

```python
df = pd.read_csv('data/student_performance.csv')
df.head()
```

## 3. Quick Data Check

```python
print(df.info())
df.describe()
```

```python
df.isna().sum()
```

## 4. Explore the Data

The first chart compares study hours with science scores.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='study_hours', y='science_score', hue='passed', palette=['#111111', '#d90429'], s=90)
plt.title('Study Hours vs. Science Score')
plt.xlabel('Study Hours')
plt.ylabel('Science Score')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='passed', y='attendance_percent', palette=['#111111', '#d90429'])
plt.title('Attendance by Pass/Fail Outcome')
plt.xlabel('Passed (0 = No, 1 = Yes)')
plt.ylabel('Attendance Percent')
plt.show()
```

## 5. Correlation Heatmap

This shows which numeric features move together.

```python
numeric_df = df.drop(columns=['student_id'])

plt.figure(figsize=(8, 6))
sns.heatmap(numeric_df.corr(), annot=True, cmap='Reds', vmin=-1, vmax=1)
plt.title('Correlation Heatmap')
plt.show()
```

## 6. Build a Simple Prediction Model

This model predicts whether a student passed using study hours, attendance, sleep, previous score, and science score.

```python
features = ['study_hours', 'attendance_percent', 'sleep_hours', 'previous_score', 'science_score']
X = df[features]
y = df['passed']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42, stratify=y)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

predictions = model.predict(X_test)
accuracy = accuracy_score(y_test, predictions)

print(f'Model accuracy: {accuracy:.2%}')
print(classification_report(y_test, predictions))
```

```python
cm = confusion_matrix(y_test, predictions)

plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Reds', xticklabels=['Fail', 'Pass'], yticklabels=['Fail', 'Pass'])
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

## 7. Findings

- Higher study hours and attendance generally connect with higher science scores.
- Lower attendance appears more common among students who did not pass.
- A simple logistic regression model can practice pass/fail prediction, but a larger real dataset would be needed for stronger conclusions.
